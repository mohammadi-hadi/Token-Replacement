```python
from contestant_model_5 import load_model
from contestant_model_5 import predict

from sklearn.metrics import f1_score, accuracy_score
import pandas as pd

best_params={'learning_rate': 6.871835594398317e-05, 'batch_size': 66, 'weight_decay': 0.000152125727160186}

# Load the dataset
csv_file_path = "/data/volume_2/CLIN/test_df.csv"  # Replace with the path to your dataset
df = pd.read_csv(csv_file_path)
text_list = df['text'].tolist()
true_labels = df['label'].tolist()

# Make predictions
predicted_labels = predict(text_list, best_params)

# Calculate F1 score and accuracy
f1 = f1_score(true_labels, predicted_labels, average='weighted')
accuracy = accuracy_score(true_labels, predicted_labels)
print(f"F1 Score for {csv_file_path}: {f1}")
print(f"Accuracy for {csv_file_path}: {accuracy:.4f}")

# Save the results
result_df = pd.DataFrame({
    'text': text_list,
    'true_label': true_labels,
    'predicted_label': predicted_labels
})

result_csv_path = "/data/volume_2/CLIN/predicted_results.csv"  # Replace with your desired save path
```
